```python
#Отобразить вектор размера 100, в котором вычеркивается x, если x ---составное
is_prime = np.ones(100, dtype = bool)
is_prime[:2] = False
```

```python
N_max = int(np.sqrt(len(is_prime)))
for i in range(2,N_max):
    is_prime[2*i::i] = False

print(is_prime)
```

5. Линейная алгебра


```python
a = np.array([[2,1],[2,3]])
print(a)
```

```python
#det
np.linalg.det(a)
```

```python
#a^-1
b = np.linalg.inv(a)
print(b)
```

```python
print(a.dot(b))
print(b.dot(a))
```

```python
c = np.array ([[2,1],[6,3]])
print(c)
print(np.linalg.det(c))
```

```python
np.linalg.inv(c)
```

```python
#Решение НЛУ
v = np.array([5,-10])
print(np.linalg.solve(a,v))

print(b.dot(v))
```

```python
#С/в и с/з
l,u = np.linalg.eig(a)
print(l)
print(u)
```

```python
l,u = np.linalg.eig(a.T)
print(l)
print(u)
```

```python
l, u = np.linalg.eig(np.eye(3))
print(l)
print(u)
```
