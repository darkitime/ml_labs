```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

####Basic graph

```python
x = [0, 1, 2, 3, 4]
y = [0, 2, 4, 6, 8]

#Resize your Graph
plt.figure(figsize=(5,4), dpi=150)

#plt.plot(x,y, label='2x', color = 'blue', linewidth = 2, marker = '.', markersize = 10, markeredgecolor = 'gray', linestyle ='--') # to get rid of annoying print

#Use shorthand notation
#fmt = '[color][marker][line]'
plt.plot(x,y, 'r.--',label='2x')

## Line number Two
x2 = np.arange(0,5,0.5)
plt.plot(x2[:5],x2[:5]**2, 'r', label ='X^2')
plt.plot(x2[4:], x2[4:]**2, 'r--')


plt.title('Our first graph', fontdict={'fontname': 'Comic Sans Ms', 'fontsize': 20})
plt.xlabel('X axis', fontdict={'fontname': 'Comic Sans Ms', 'fontsize': 20})
plt.ylabel('Y axis')

plt.xticks([0,1,2,3,4])
plt.yticks([0,2,4,6,7, 7.5,10])

plt.legend()

plt.savefig('mygraph.png', dpi = 300)

plt.show()
```

#### Bar Chart

```python
labels = ['A', 'B', 'C']
values = [1, 4, 2]

bars = plt.bar(labels, values)

patterns = ['/', 'O', '*']
for bar in bars:
    bar.set_hatch(patterns.pop(0))
# bars[0].set_hatch('/') 
# bars[1].set_hatch('O')
# bars[2].set_hatch('*')

plt.figure(figsize=(6,4))

plt.show()
```

### Fifa data


```python
fifa = pd.read_csv('fifa_data.csv')
fifa.head()
```

### Histograms

```python
bins = [i for i in range(40,101,10)]


plt.hist(fifa.Overall, bins=bins, color='#abcdef' )
plt.xticks(bins)

plt.title('Distib of players skills in FIFA 2018')
plt.xlabel('Skill level')
plt.ylabel('Number of players')

plt.show()
```

##Pie Charts

```python
left = fifa.loc[fifa['Preferred Foot'] == 'Left'].count().iloc[0]
right = fifa.loc[fifa['Preferred Foot'] == 'Right'].count().iloc[0]

labels = ['Left', 'Right']
colors = ['#abcdef','#aabbcc']

plt.pie([left,right], labels=labels, colors = colors, autopct='%.2f %%')
plt.title(f'Foot Preferemce of FIFA Players')

plt.show()
```

```python
fifa.Weight = [int(x.strip('lbs')) if type(x) == str else x for x in fifa.Weight]
fifa.Weight

```

```python
plt.style.use('ggplot')

light = fifa.iloc[fifa.Weight < 125].count().iloc()[0]
light_med = fifa.iloc[(fifa.Weight>=125)&(fifa.Weight<150)].count().iloc()[0]
medium = fifa.iloc[(fifa.Weight>=150)&(fifa.Weight<175)].count().iloc()[0]
medium_heavy = fifa.iloc[(fifa.Weight>=175)&(fifa.Weight<200)].count().iloc()[0]
heavy = fifa.iloc[(fifa.Weight>=200)].count().iloc()[0]

weights = [light, light_med, medium, medium_heavy, heavy]
labels = ['Under 125', '125-150','150-175','175-200','Over 200']
explode = (.4,.1,0,0,0.4)
plt.pie(weights, labels=labels, autopct='%.2f %%', pctdistance=0.8, explode = explode)

plt.title('Weight Distribution of FIFA Players (in lbs)')



plt.show()
```

```python
plt.style.use('default')

barcelona = fifa.iloc[fifa.Club == 'FC Barcelona']['Overall']
madrid = fifa.iloc[fifa.Club == 'Real Madrid']['Overall']

labels = ['FC Barcelona', 'Real Madrid']

boxes = plt.boxplot([barcelona, madrid], tick_labels = labels, patch_artist= True, manage_ticks=True, medianprops={'linewidth':2})

for box in boxes['boxes']:
    box.set(color='#4286f4', linewidth=2)
    box.set(facecolor='#e0e0e0')


plt.title('Professional Football Team Comparison')
plt.ylabel('FIFA Overall Rating')

plt.show()
```
