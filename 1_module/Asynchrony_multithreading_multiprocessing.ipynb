```python
from tkinter import ttk
import tkinter as tk
import time
from threading import Thread

root = tk.Tk()
root.geometry('300x300+150+150')

def thread_sleep():
    time.sleep(10)
    lab['text'] = int(lab['after sleep']) + 10

def new_thread():
    t1 = Thread(target=thread_sleep)
    t1.start()

btn = ttk.Button(root, text = 'Run', command = new_thread) #Запуск нового потока, в котором работает time.sleep
btn.place(relx=0.5, rely=0.2,anchor=tk.CENTER)

lab = ttk.Label(root, text = 'before click')
lab.place(relx = 0.5, rely =0.6, anchor = tk.CENTER)

root.mainloop()
#В чем разница? Процесс - отдельно запущенная независимая программа. Поток - ест ресурсы процесса
#Питон заточен под однопоточность
#Если идет ожидание, то глобальный интерпретатор передает ресурс тому, кто не спит
#сколько потоков оптимально 4-40: при дальнейшем увеличении при переключении тратится время и уже не выгодно
```

```python
import time
import os

def clock():
    time0 = round(time.time())
    while True:
        if (round(time.time()) - time0) % 5 == 0:
            yield ('5 sec')
        else:
            yield 0

def query():
    for i in os.walk('C:\\'):
        yield i[0] #делаем генератор функцией yield - возвращает объект ожилания
#мжно for\next получать значения
#x = query()
#print(x)

def main():
    data = query()
    alarm = clock() #clock не запустится до завершения 
    while True: #Две функции работают асинхронно
        d = next(data)
        a = next(alarm)
        print(d)
        if a:
            print(a)
            time.sleep(1)

main()
#next(x)
#next(x)
#можно использовать генераторы для создания асинхронного кода
```

```python
import asyncio
from time import time

async def print1(sec):
    await asyncio.sleep()
    print(sec)

async def main():
    async with asyncio.TaskGroup() as tg:
        for i in range(1,16):
            tg.create_task(print1(i))


start = time()
await main()
print(f"Time: {start}")
```
