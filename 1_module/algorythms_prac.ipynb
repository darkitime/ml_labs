```python
nums = [9,12,5,10,14,3,10]
pivot = 10
def pivotArray(nums:list[int],pivot:int)->list[int]:
    result1=[]
    result2=[]
    result3=[]
    for i in range(len(nums)):
        if nums[i]<pivot:
            result1.append(nums[i])
        elif nums[i]>pivot:
            result2.append(nums[i])
        else:
            result3.append(pivot)
    return (result1+result3+result2)

print(pivotArray(nums,pivot))
```

```python
n = "32"
def minPartitions(n:str)->int:
    max_s=0
    for s in n:
        max_s=max(max_s,int(s))
    return max_s
        

print(minPartitions(n))
```

```python
boxes = "001011"
def minOperations(boxes: str)->list[int]:
    result = [0]*len(boxes)
    for i in range(len(boxes)):
        for j in range(len(boxes)):
            if boxes[j]=='1':
                result[i]+= abs(i-j)
    return result
boxes = "110"
print(minOperations(boxes))
```
