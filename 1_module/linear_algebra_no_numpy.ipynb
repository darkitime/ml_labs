```python
matrix = [[4, 7], [2, 6]]

def inverse_2x2(matrix: list[list[float]]) -> list[list[float]] | None:
    a=int(matrix[0][0])
    b=int(matrix[0][1])
    c=int(matrix[1][0])
    d=int(matrix[1][1])
    det = a*d-b*c
    
    if det == 0:
        return None
    inversed_matrix = [[d*(1/det) ,-b*(1/det) ],[-c*(1/det) ,a*(1/det) ]]
    return inversed_matrix

print(inverse_2x2(matrix))
```

```python
B = [[1, 0, 0], 
    [0, 1, 0], 
    [0, 0, 1]]

C = [[1, 2.3, 3], 
    [4.4, 25, 6], 
    [7.4, 8, 9]]

def transform_basis(B: list[list[int]], C: list[list[int]]) -> list[list[float]]:
    if len(C) == 3:
        m=C
        det = (
            m[0][0] * (m[1][1] * m[2][2] - m[1][2] * m[2][1])
            - m[0][1] * (m[1][0] * m[2][2] - m[1][2] * m[2][0])
            + m[0][2] * (m[1][0] * m[2][1] - m[1][1] * m[2][0])
        )
        if det == 0:
            return None
        matrix = C
        cofactors = [
            [
                +(matrix[1][1] * matrix[2][2] - matrix[1][2] * matrix[2][1]),
                -(matrix[1][0] * matrix[2][2] - matrix[1][2] * matrix[2][0]),
                +(matrix[1][0] * matrix[2][1] - matrix[1][1] * matrix[2][0]),
            ],
            [
                -(matrix[0][1] * matrix[2][2] - matrix[0][2] * matrix[2][1]),
                +(matrix[0][0] * matrix[2][2] - matrix[0][2] * matrix[2][0]),
                -(matrix[0][0] * matrix[2][1] - matrix[0][1] * matrix[2][0]),
            ],
            [
                +(matrix[0][1] * matrix[1][2] - matrix[0][2] * matrix[1][1]),
                -(matrix[0][0] * matrix[1][2] - matrix[0][2] * matrix[1][0]),
                +(matrix[0][0] * matrix[1][1] - matrix[0][1] * matrix[1][0]),
            ],
        ]

        adjugate = [
            [cofactors[j][i] for j in range(3)] 
            for i in range(3)
        ]


        inv_matrix = [
            [round(adjugate[i][j] / det, 6) for j in range(3)] 
            for i in range(3)
        ]

    else:
        a=int(C[0][0])
        b=int(C[0][1])
        c=int(C[1][0])
        d=int(C[1][1])
        det = a*d-b*c
            
        if det == 0:
            return None
        inv_matrix = [[d*(1/det) ,-b*(1/det) ],[-c*(1/det) ,a*(1/det) ]]
        

    P = [[0 for _ in range(len(B[0]))] for _ in range(len(B))]

    for row in range(len(B)):
        for col in range(len(B[0])):
            P[row][col] = sum(inv_matrix[row][k]*B[k][col] for k in range(len(B)))

    return P

print(transform_basis(B,C))
```

```python
matrix=[1,2,3]
def make_diagonal(matrix:list[int])->list[list[int]]:
    diagonal_matrix = [[0 for _ in range(len(matrix))] for _ in range(len(matrix))]
    for i in range (len(matrix)):
        for j in range(len(matrix)):
            if i==j:
                diagonal_matrix [i][j] = int(matrix[i])
            else:
                diagonal_matrix [i][j] = 0

    return diagonal_matrix

print(make_diagonal(matrix))
```

```python
dense_matrix = [
    [1, 0, 0, 0],
    [0, 2, 0, 0],
    [3, 0, 4, 0],
    [1, 0, 0, 5]
]
def compressed_row_sparse_matrix(dense_matrix):
    vals = []
    col_idx = []
    row_ptr = []

    for row in range(len(dense_matrix)):
        for col in range(len(dense_matrix[0])):
            if dense_matrix[row][col]!=0:
                vals.append(dense_matrix[row][col])
                col_idx.append(col)
        row_ptr.append(len(vals))
        

    return vals, col_idx, row_ptr


vals, col_idx, row_ptr = compressed_row_sparse_matrix(dense_matrix)
print("Values array:", vals)
print("Column indices array:", col_idx)
print("Row pointer array:", row_ptr)
```

```python
v = [3, 4]
L = [1, 0]
def orthogonal_projection(v, L):
    len_L_square=sum(L[i]**2 for i in range(len(L)))
    solcar_multiply = sum(v[i]*L[i] for i in range(len(v)))
    result = [solcar_multiply*L[i]/len_L_square for i in range(len(v))]
    return result
print(orthogonal_projection(v, L))
```

```python
v1=[1,2,3]
v2=[2,4,6]

def cosine_similarity(v1, v2):
    solar_multy = sum(v1[i]*v2[i] for i in range(len(v1)))
    abs_v1 = (sum(v1[i]**2 for i in range(len(v1))))**(1/2)
    abs_v2 = (sum(v2[i]**2 for i in range(len(v2))))**(1/2)
    cosine_similarity = solar_multy/(abs_v1*abs_v2)
    return cosine_similarity

print(cosine_similarity(v1,v2))
```

```python
data = [1.0, 2.0, 3.0] # число в начале определяет число в конце. Если лен 2 до будет два подмассива. Если три, то будет три подмассива
degree = 0 #Количество элементов в массие
def phi_transform(data: list[float], degree: int) -> list[list[float]]:
    if degree == 0:
        return []
    
    result=[]
    for elem in data:
        sub_data = [elem**i for i in range(0,degree+1)]
        result.append(sub_data)
    return result

print(phi_transform(data,degree))

```
