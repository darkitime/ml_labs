```python
#Пересекающиеся окна
my_lst = [1,0,1,0,1,0,1,1]
k=2

def longest_ones_with_flips(nums: list[int], k:int) -> int:
    #Initialising parametrs
    l = 0
    r = -1
    result = 0

    #Window status
    count_zeros = 0

    while l<len(nums):
        #Расширяем окно пока может
        while r+1 < len(nums) and (nums[r+1] == 1 or count_zeros < k): #Условие расширения
            #Обновление состояния
            if nums[r+1] == 0:
                count_zeros+=1
            r+=1
        #Обновление результата
        result = max(result, r-l+1)
        #Обновление состояния
        if nums[l] == 0:
            count_zeros -=1
        l+=1
    return result

print(longest_ones_with_flips(nums=my_lst, k=k))
#o(n) - time, o(1)-memory 
#Саммая длинная последовательность с пересекающимися окнами
```

```python
#Паттерн: Метод отрезков

segments = [[2,4],[1,3],[5,6],[6,7]]

def unification_segments(segments: list[list[int]]) -> list[list[int]]:
    #Сортировка
    segments.sort()
    #Инициализация
    result = [segments[0]]
    index=0
    for i in range (1, len(segments)):
        if max(result[index][0],segments[i][0])<=min(result[index][1],segments[i][1]):
            result[index] = [min(result[index][0],segments[i][0]),max(result[index][1],segments[i][1])]
        else:
            result.append(segments[i])
            index+=1
    return result

print(unification_segments(segments = segments))

#Второй вариантт

def is_overlapping(a:list[int],b:list[int]) -> bool:
    return max(a[0],b[0]) <= min(a[1],b[1])

def merge_two (a: list[int], b:list[int]) -> list[int]:
    return [min(a[0],b[0]), max(a[1],b[1])]

def merge (segments: list[list[int]]) -> list[list[int]]:
    segments.sort()

    result = [segments[0]]

    for i in range(1,len(segments)):
        last = result[-1]
        current = segments[i]
        if is_overlapping(last,current):
            result[-1] = merge_two (last, current)
        else:
            result.append(current)
    return result

print(merge(segments=segments))
#O(N*logN), O(N)
#Нужно объединить, найти пересечения в одном списке отрезков или временных промежутков
```

```python
#Указатели на отрезках
segments1 = [[1,2], [3,4], [6,7]]
segments2 = [[0,5], [7,8]]

def overlap_two_segments(a:list[int],b:list[int]) -> list[int]:
    return [max(a[0], b[0]), min(a[1], b[1])]

def merges_two_segments(segments1: list[list[int]], segments2: list[list[int]]) -> list[list[int]]:
    p1 = 0
    p2 = 0
    result = []
    while p1 < len(segments1) and p2 < len(segments2):
        if is_overlapping(a=segments1[p1], b=segments2[p2]):
            result.append(overlap_two_segments(a=segments1[p1], b=segments2[p2]))

        if segments1[p1][1] < segments2[p2][1]:
            p1 += 1
        else: 
            p2+=1
    return result

print(merges_two_segments(segments1=segments1,segments2=segments2))
#O(N+M), O(max(N,M))
#встречаем несколько списков интервалов

```

```python
#стек
stack = ['(',")","<","[","{","}","]",">"]

def  is_valid(s:str) -> bool:
    #инициализация
    stack = []
    mapping = {
        '(':")",
        "<": ">",
        "[": "]",
        "{":"}"
    }

    for char in s:
        if char in mapping:
            #char - открывающая скобка
            stack.append(char)
        elif len(stack) == 0:
            #char - закрывающая скобка и стек пуст
            return False
        else:
            #char - закрывающая скобка и стек не пуст
            last_char = stack.pop()
            if mapping[last_char] != char:
                return False
    return len(stack)==0
    
print(is_valid(stack))

#O(N), O(N)
#Вычислить выражение, нужно работать со скобками или вложенными структурами 
```

```python
#Односвязный список
class Node:
    def __init__(self,data,next) :
        self.data=data
        self.next = next

head = Node(1,Node(2,Node(3,None)))
        
```

```python
#1)Счетчик вершин

def size(head:Node):
    count = 0
    curr = head
    while curr:
        count+=1
        curr = curr.next
    return count

print(size(head))
#O(n),O(1)
```

```python
#2)Середина
def middle(head:Node) -> Node:
    slow = head
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    return slow

print(middle(head).data)
```

```python
#3)Разворот
def reverse(head:Node) -> Node:
    prev = None
    curr = head
    while curr:
        tmp = curr
        curr = curr.next
        tmp.next = prev 
        prev=tmp
    return prev

print(reverse(head))
#O(n), O(1)
```

```python
#Паттерн - фиктивный узел
head1 = Node(1,Node(2,Node(3,None)))
head2 = Node(2,Node(2,Node(7,Node(8,None))))

def merge_nodes(head1:Node,head2:Node)->Node:
    #создание dummy
    dummy = Node(0, None)

    #формирование ответа
    curr = dummy
    while head1 and head2:
        #каждому по указателю
        if head2 is None or (head1 is not None and head1.data<=head2.data):
            curr.next = head1
            head1 = head1.next
        else:
            curr.next = head2
            head2 = head2.next
        curr = curr.next
    return dummy.next

print(merge_nodes(head1,head2))
#O(N+M), O(1)
#нужен узел, чтобы крепить к нему ответ
#хотим обрабатывать удаление первого узла из списка так же как остальные
```

```python
#Перебор комбинаций
#1)брутфорс
from collections import deque

def generate_combinations(s: str) -> list[str]:
    phone_map = {
        "2": "abc", "3": "def", "4": "ghi", "5": "jkl",
        "6": "mno", "7": "pqrs", "8": "tuv", "9": "wxyz"
    }
    #Инициализация очереди\дека
    result = deque([""])
    #цикл с условием остановки перебора
    while result and len(result[0])!=len(s):
        prefix = result.popleft()
        digit = s[len(prefix)]

        for letter in phone_map[digit]:
            result.append(prefix+letter)
    return list(result)

print(generate_combinations("2469"))
#O(N*4^N)
#Генерация всех возможных комбинаций
         
```

```python
#бэктрэкинг
from collections import deque

def generate_brackets(n: int) -> list[str]:
    #инициализация дека: один элемент — кортеж (строка, открытых, закрытых)
    q = deque([("", 0, 0)])
    #цикл с условием остановки перебора
    while q and len(q[0][0]) < 2 * n:
        #Достаем первый элемент очереди
        prefix, open_used, close_used = q.popleft()
        #Генерация валидных комбинаций с вставкой в конец очереди
        if open_used < n:
            q.append((prefix+"(", open_used +1, close_used))
        if close_used < open_used:
            q.append((prefix+")",open_used,close_used+1))
    return [prefix for prefix,_,_ in q]

print(generate_brackets(5))   
#O(2^2N),O(2^2N)~O(sqrt(n)*4^N)
#Генерация последовательности с правилами
```

```python
#Деревья 
class TreeNode:
    def __init__(self, val=0, left=None, right = None):
        self.val = val
        self.left = left
        self.right = right

root = TreeNode(5)

root.left = TreeNode(8)
root.right = TreeNode(1)

root.left.left = TreeNode(4)

root.right.left = TreeNode(4)
root.right.right = TreeNode(1)

```

```python
#1)Снизу -> вверх
#Число узлов


def count_nodes(node: TreeNode) -> int:
    if node is None:
        return 0
    left = count_nodes(node.left)
    right = count_nodes(node.right)

    return left+right+1

print(count_nodes(root))
#O(N), O(h)
```

```python
def find_diameter(node: TreeNode)-> int:
    diameter = 0
    def dfs(node: TreeNode) -> int:
        #base
        if node is None:
            return 0
        left = dfs(node.left)
        right = dfs(node.right)

        nonlocal diameter
        diameter = max(diameter, left+right)
        return max(left,right)+1
    dfs(node)
    return diameter

print(find_diameter(root))
#o(N), o(h)
#Сбор ответа снизу вверх
```

```python
#Сверху вниз
target_sum = 15

def is_way(root:TreeNode, target_sum: int) ->True:
    def dfs(node: TreeNode, curr_sum: int) -> bool:
        #base
        if node is None:
            return False

        curr_sum+=node.val
        if node.left is None and node.right is None:
            return curr_sum == target_sum

        return dfs(node.left,curr_sum) or dfs(node.right,curr_sum)

    return dfs(root,0)

print(is_way(root,target_sum))
#O(N), O(h)
```

```python
from collections import deque 
 
  #Поиск в ширину
def shortest_path(graph: dict[int,list[int]], start: int, end:int) -> int:
    queue = deque([(start,0)])
    visited = {start}

    while queue:
        node, dist = queue.popleft()

        if node == end:
            return dist
        
        for neighbor in graph[node]:
            if neighbor not in visited:
                queue.append((neighbor,dist+1))
                visited.add(neighbor)
            
    return -1

print(shortest_path(graph_from_edges(edges),1,2))
#O(N+M), O(N)
```

```python
#ООП
#Одномерная динамика
N=5

def climb_stairs(n:int)->int:
    #inisialisation
    dp = [0]*(n+1)

    #base
    dp[0] = 1
    dp[1] = 1

    for i in range(2, n+1):
        dp[i] = dp[i-1]+dp[i-2]

    return dp[n]

print(climb_stairs(N))
#O(N),O(1)
```

```python
#Двумерная динамика
def knapsack(weights: list[int], values: list[int],capacity: int)->int:
    #Инициализация
    dp = [0] * (capacity + 1)

    for i in range(len(weights)):
        weight = weights[i]
        value = values[i]

        for w in range (capacity, weight-1, -1):
            dp[w] = max(dp[w],dp[w-weight] + value)
    return dp[capacity]

print(knapsack([6,3,2],[6,6,6],6)) #O(N*W), O(W)
```
